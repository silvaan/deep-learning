# Ranking da competição — Beans

```python
from pathlib import Path
import pandas as pd

gt = pd.read_csv("data/ground_truth.csv")["Class"]
gt.value_counts()
```

```python
rows = []
for f in sorted(Path("submissoes/competition").glob("*.csv")):
    pred = pd.read_csv(f)["Class"].astype(str).str.strip()
    hits = int((pred.values == gt.values).sum()) if len(pred) == len(gt) else 0
    rows.append({"aluno": f.stem, "linhas": len(pred), "acertos": hits, "acurácia": hits / len(gt)})

ranking = pd.DataFrame(rows).sort_values("acurácia", ascending=False, ignore_index=True)
ranking.index += 1
ranking.style.format({"acurácia": "{:.2%}"})
```
